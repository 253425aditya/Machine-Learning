```python
# Toss winner is Match winner in percentage 
(ipl[ipl['TossWinner'] == ipl['WinningTeam']].shape[0]/ipl.shape[0])*100
```

```python
# write a function to return the track record of  2 teams against each other 
def against(team1, team2):
    temp = ipl[((ipl['Team1'] == team1) | (ipl['Team2'] == team1)) & ((ipl['Team1'] == team2) | (ipl['Team2'] == team2))]
    match_count = temp.shape[0]
    print(temp[["Team1","Team2","WinningTeam"]])
    win_team1 = temp[temp['WinningTeam'] == team1].shape[0]
    print(f"Total matches : {match_count}")
    print(f"{team1} wins : {win_team1}")
    print(f"{team2} wins : {match_count-win_team1}")
    print(f"If {team1} play against {team2}, Winning chance of {team1} is {round((win_team1/match_count)*100,2)}")

against('Rajasthan Royals','Deccan Chargers')
```
