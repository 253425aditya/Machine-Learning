# Seires from Dict


```python
marks_dict = {
    'Maths' : 45,
    'English' : 85, 
    'Science' : 77, 
    'Cs' : 86,
    'Hindi' : 99
}
marks_series = pd.Series(marks_dict, name="Aditya marks")
marks_series
```
