# Editing Series

```python
marks = pd.Series({'Maths' : 89,'Science' : 87, 'English' : 77, 'Hindi' : 96})
marks
```

```python
# Using indexing
marks['English'] = 100
marks
```

```python
marks['Physics'] = 67
marks
```

```python
# using Slicing
marks.iloc[2:4] = [50, 50]
```

```python
marks
```

```python
# using Fancy indexing
marks.iloc[[2,4]] = [60,78]
marks
```

```python
marks[marks>70]
```
