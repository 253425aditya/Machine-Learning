# Slicing

```python
import numpy as np
```

```python
arr1 = np.arange(1,10)
arr1
```

```python
arr2 = np.arange(1,25)
arr2
```

```python
arr2 = arr2.reshape(2,12)
arr2
```

```python
arr2 = arr2.reshape(6,4)
arr2
```

```python
arr2[2]
```

```python
arr2[:,:2]
```

```python
arr2[2:4,1:3]
```

```python
for i in arr2:
    print(i)
```

```python
for i in np.nditer(arr2):
    print(i)
```
