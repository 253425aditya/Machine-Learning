# Reshaping

```python
import numpy as np
```

```python
arr1 = np.arange(1,10)
arr1
```

```python
arr1 = arr1.reshape(3,3)
arr1
```

```python
arr1.ravel()
```

```python
for i in arr1:
    print(i)
```

```python
for i in arr1.ravel():
    print(i)
```

```python
arr1.transpose()
```

```python
arr2 = np.arange(10,19).reshape(3,3)
arr2
```

```python
np.hstack((arr1,arr2))
```

```python
np.vstack((arr1,arr2))
```

```python
np.hsplit(arr1,3)
```

```python
np.vsplit(arr1,3)
```

```python
arr3 = np.arange(24).reshape(6,4)
arr3
```

```python
arr3[[0,2,4]]
```

# Flitering

```python
arr4 = np.random.randint(low=1, high=100,size=20).reshape(4,5)
arr4
```

```python
arr4>30
```

```python
arr4[arr4>50]
```

```python
arr4[(arr4>50) & (arr4%2!=0)]
```

```python
arr4[arr4<30] = 0
arr4
```
