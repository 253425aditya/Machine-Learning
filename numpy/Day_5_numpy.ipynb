# Broadcasting


```python
import numpy as np 
```

```python
arr1 = np.arange(8).reshape(2,4)
arr2 = np.arange(8,16).reshape(2,4)
print(arr1)
print(arr2)
```

```python
arr1+arr2
```

```python
arr1+[10]
```

```python
arr3 = np.arange(9).reshape(3,3)
arr4 = np.arange(3).reshape(1,3)
print(arr3)
print(arr4)
```

```python
arr3+arr4
```

```python
arr4 = arr4.reshape(3,1)
arr4
```

```python
arr3+arr4
```
